Define a function named **count_letters**.

It will ask for sentence from the user.

And it will count the occurences of each letter and print it.

**Hints:**
* Do not count space character
* Use the function you defined in Q2.

<pre>
Expected Output:

Please enter a sentence:  pandas
letter p: 1 time(s)
letter a: 2 time(s)
letter n: 1 time(s)
letter d: 1 time(s)
letter s: 1 time(s)
</pre>

```python
# S 3:

def count_letters():
    user_input_str = input("Please enter a sentence: ")
    save = "";
    for i in user_input_str:
        if i == " ":
            continue
        save += i;
        amount = count_the_letter(user_input_str, i)
        if (save.count(i) < 2):
            print("letter {0}: {1} time(s)".format(i, amount))
        
        
```

```python
count_letters()
```

**Q 8:**

Define a function named **split_email**.

It will ask for an email from the user.

And it will split the email into user name and domain name using **index**.

**Hints:**
* if the input doesn't include '@' char, return a text as 'Invalid email format.'

<pre>
Expected output:

Please enter an email:  abc@acme.corp
User Name: abc
Domain: acme.corp
</pre>

```python
# S 8:

def split_email():
    user_input = input("Please enter an email: ")
    if user_input.find("@") == -1:
        return "Invalid email format"
    user_name = user_input[:user_input.find("@")]
    domain_name = user_input[user_input.find("@") + 1:]
    print("User Name: {0}".format(user_name))
    print("Domain: {0}".format(domain_name))
    
```

```python
split_email()
```

**Q 10:**

Define a function named **replace_with_next_letter**.

It will ask for an input from the user.

And will replace every character with the next letter in the English alphabet.

**Hints:**
* use nested functions
* there should be is_lower() and is_upper() functions to check whether the char is small or upper
* there should be next_letter() function to take a letter and return the next one from alphabet
* alphabets:
    * alphabet_lower = 'abcdefghijklmnoprstuvyz'
    * alphabet_upper = 'ABCDEFGHIJKLMNOPRSTUVYZ'
* use index to get the related letter

<pre>
Expected Output:

Please enter a text:  Python
Rzuipo

Please enter a text:  abcZ
bcdA
</pre>

```python
# S 10:

def is_lower(char):
    alphabet_lower = 'abcdefghijklmnoprstuvyz'
    if char in alphabet_lower:
        return True
    return 

def is_upper(char):
    alphabet_upper = 'ABCDEFGHIJKLMNOPRSTUVYZ'
    if char in alphabet_upper:
        return True
    return False

def next_letter(char):
    alphabet_lower = 'abcdefghijklmnoprstuvyz'
    alphabet_upper = 'ABCDEFGHIJKLMNOPRSTUVYZ'
    if is_lower(char):
        if alphabet_lower[alphabet_lower.find(char)] == alphabet_lower[-1]:
            return alphabet_lower[0]
        return alphabet_lower[alphabet_lower.find(char) + 1]
    elif is_upper(char):
        if alphabet_upper[alphabet_upper.find(char)] == alphabet_upper[-1]:
            return alphabet_upper[0]
        return alphabet_upper[alphabet_upper.find(char) + 1]

def replace_with_next_letter():
    user_input = input("Please enter a text: ")
    new_text = ""
    for char in user_input:
        new_text += next_letter(char)
    return new_text
        

    
```

```python
new_text = replace_with_next_letter()
print(new_text)
```
